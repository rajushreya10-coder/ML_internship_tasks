## Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Explore Basic Info


```python
ds = pd.read_excel('1.xlsx')
print(ds.shape)
print(ds.info())
print(ds.describe())
print(ds.head())
print(ds.isnull().sum())
print(ds.columns)
print(ds.dtypes)
```

## Dataset Cleaning

```python
for col_name in ['gender', 'country', 'residence', 'prevEducation']:
    ds[col_name] = ds[col_name].str.lower().str.strip()
```

```python
X = ds.iloc[: , 2:9].values
y = ds.iloc[: , 9:].values
```

```python
print(X)
```

```python
print(y)
```

```python
from sklearn.impute import SimpleImputer
im = SimpleImputer(missing_values=np.nan , strategy = 'mean' )
im.fit(y[: , 0:1 ])
y[: , 0:1 ] = im.transform(y[: , 0:1 ])
```

```python
print(y)
```

## Encoding

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(sparse_output=False), [1, 2, 3, 5])],
    remainder='passthrough'
)

X = ct.fit_transform(X)

ds_result = pd.DataFrame(X)
pd.set_option('display.max_columns', None)
print(ds_result)
```

## Splitting dataset

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

## Visualisation


```python
numerical_cols = ['Age', 'entryEXAM', 'studyHOURS']

plt.figure(figsize=(8, 5))
ds.boxplot(column=numerical_cols)
plt.title('Boxplots of INPUT Features (Before Cleaning)')
plt.show()
```

## Removing Outliers

```python
print(f"Original shape of data: {ds.shape}")


Q1 = ds[numerical_cols].quantile(0.25)
Q3 = ds[numerical_cols].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

mask = (ds[numerical_cols] >= lower_bound) & (ds[numerical_cols] <= upper_bound)

ds_cleaned = ds[mask.all(axis=1)]

print(f"Cleaned shape of data: {ds_cleaned.shape}")
```

```python
plt.figure(figsize=(8, 5))

ds_cleaned.boxplot(column=numerical_cols)
plt.title('Boxplots of INPUT Features (AFTER Cleaning)')
plt.show()
```
